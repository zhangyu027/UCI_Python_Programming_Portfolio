## Problem 1
Determine the value of $i^i$ as a real number, where $i=\sqrt{-1}$.

```python
import math

# i = e^(i*pi/2), so i^i = (e^(i*pi/2))^i = e^(-pi/2)
answer_p1 = math.exp(-math.pi / 2)
answer_p1
```

## Problem 2
The top of the die is given by $3(a^3b - ab^3) \bmod 7$. Compute for (a) a=2, b=6 and (b) a=3, b=5.

```python
def die_top(a, b):
    return (3 * (a**3 * b - a * b**3)) % 7
```

```python
case_a = die_top(2, 6)
case_b = die_top(3, 5)
case_a, case_b
```

## Problem 3
A sheet of paper has thickness $t=0.1$ mm. How many folds are needed to reach the Moon, distance $d=384,400$ km? After $n$ folds, thickness is $2^n t$.

```python
import math

t_mm = 0.1
d_km = 384_400
d_mm = d_km * 1_000_000  # 1 km = 1,000,000 mm

n = math.ceil(math.log2(d_mm / t_mm))
# thickness_after_n_mm = (2 ** n) * t_mm
# thickness_after_n_km = thickness_after_n_mm / 1_000_000
n
# n, thickness_after_n_km
```

## Problem 4
Explain why `10^2` does not evaluate to 100.

```python
# In Python, ^ is bitwise XOR, not exponentiation.
# 10 in binary is 1010; 2 in binary is 0010.
# XOR gives 1000 in binary, which is 8.
wrong = 10 ^ 2
correct = 10 ** 2
wrong, correct
```

## Problem 5
Slice the string `s = 'seehemewe'` to produce the requested substrings.

```python
s = 'seehemewe'

answers_p5 = {
    "see": s[0:3],
    "he": s[3:5],
    "me": s[5:7],
    "we": s[7:9],
    "hem": s[3:6],
    "meh": s[6:3:-1],
    "wee": s[7::-3],## 7:: start to end, and -3 is the first "e" occured
}
answers_p5
```

## Problem 6
Write a single-line expression for determining if a string is a palindrome.

```python
# Single-line palindrome expression:
# Write a single-line expression for determining if a string is a palindrome (reads the same forwards as backwards)
# s == s[::-1]

for word in ['banana', 'deified', 'noon', 'civic', 'racecar', 'level', 'mom']:
    print(word, '->', word == word[::-1])
```

## Problem 7
Explain how the incorrect tests are interpreted by Python and give a correct alternative.

```python
s = 'eggs'

# Incorrect examples:
example_1 = s == ('eggs' or 'ham')   # ('eggs' or 'ham') evaluates to 'eggs'
example_2 = s == ('ham' or 'eggs')   # ('ham' or 'eggs') evaluates to 'ham'

# Correct alternatives:
correct_1 = (s == 'eggs') or (s == 'ham')
correct_2 = s in ('eggs', 'ham')

example_1, example_2, correct_1, correct_2
```
