Problem 1: scalar expressions.

```python
a = 1.12
b = 2.34
c = 0.72
d = 0.81
f = 19.83

x = a / b + c / f**2
s = (b - a) / (d - c)
r = 1 / (1/a + 1/b + 1/c + 1/d)
y = (a * b * c) / (2 * f**2)

print('x =', x)
print('s =', s)
print('r =', r)
print('y =', y)
```

Problem 2: complex-number expressions.

```python
import cmath
x = 5 + 8j
y = -6 - 7j

u = x + y
v = x * y
w = x / y
z = cmath.exp(x)
r = cmath.sqrt(y)
s = x * y**2

print('u =', u)
print('v =', v)
print('w =', w)
print('z =', z)
print('r =', r)
print('s =', s)
```

Problem 3: complex arithmetic.

```python
print('(3+6j)*(-7-9j) =', (3 + 6j) * (-7 - 9j))
print('(5+4j)/(5-4j) =', (5 + 4j) / (5 - 4j))
print('3/(2j) =', 3 / (2j))
```

Problem 4: NumPy calculations. The assignment says log means base 10.

```python
import numpy as np

a = np.exp(-2.1)**3 + 3.47 * np.log10(14) + 287**(1/4)
b = (3.4)**7 * np.log10(14) + 287**(1/4)
c = np.cos(4.12 * np.pi / 6)**2
d = np.cos((4.12 * np.pi / 6)**2)

print('a =', a)
print('b =', b)
print('c =', c)
print('d =', d)
```

Problem 5: Boolean expressions for x = 6.

```python
x = 6
print('x < 10  =', x < 10)
print('x == 10 =', x == 10)
print('x >= 4  =', x >= 4)
print('x != 7  =', x != 7)
```

Problem 6: Boolean/precedence checks.

```python
expressions = {
    '6 > 3 + 8': 6 > 3 + 8,
    '6 + 3 > 8': 6 + 3 > 8,
    '4 > (2 + 9)': 4 > (2 + 9),
    '(4 < 7) + 3': (4 < 7) + 3,
    '4 < 7 + 3': 4 < 7 + 3,
    '(4 < 7) * 5': (4 < 7) * 5,
    '4 < (7 * 5)': 4 < (7 * 5),
    '2/5 >= 5': 2/5 >= 5,
}
for expression, result in expressions.items():
    print(f'{expression} = {result}')
```

Problem 7: string expressions.

```python
A = 'Data science professionals have promising career path'

print('A[8] =', A[8])
print('A[-1] =', A[-1])
print('A[-6] =', A[-6])
print('A[3:10] =', A[3:10])
print("'promising' in A =", 'promising' in A)
print("A.lower().replace('career path','future').lstrip('d') =", A.lower().replace('career path', 'future').lstrip('d'))
print("A.index('science') =", A.index('science'))
```
