Problems 5–10 use the matrix shown in the assignment.

```python
import numpy as np
A = np.array([[3, 7, -4, 12],
              [-5, 9, 10, 2],
              [6, 13, 8, 11],
              [15, 5, 4, 1]])
print(A)
```

Problem 5.

```python
v = A[:, 1]
w = A[1, :]
print('v = second column:', v)
print('w = second row:', w)
```

Problem 6.

```python
B = A[:, 1:4]
C = A[1:4, :]
D = A[0:2, 1:4]
print('B =\n', B)
print('C =\n', C)
print('D =\n', D)
```

Problem 7.

```python
x = np.array([2, 4, 7])
y = np.array([2, -4, 7])
z = np.array([5 + 3j, -3 + 4j, 2 - 7j])
for name, vector in [('x', x), ('y', y), ('z', z)]:
    print(name, 'length =', len(vector))
    print(name, 'absolute values =', np.abs(vector))
```

Problem 8.

```python
print('Column maximums:', A.max(axis=0))
print('Column minimums:', A.min(axis=0))
print('Row maximums:', A.max(axis=1))
print('Row minimums:', A.min(axis=1))
```

Problem 9.

```python
B = np.sort(A, axis=0)  # sort each column
C = np.sort(A, axis=1)  # sort each row
D = A.sum(axis=0)       # add each column
E = A.sum(axis=1)       # add each row
print('B =\n', B)
print('C =\n', C)
print('D =', D)
print('E =', E)
```

Problem 10.

```python
A2 = np.array([[1, 4, 2],
               [2, 4, 100],
               [7, 9, 7],
               [3, np.pi, 42]], dtype=float)
B2 = np.log(A2)

print(B2)
```

```python
print('A Matrix =\n', A2)
print('B Matrix =\n', B2)
```

```python
a = B2[1, :] #second row of B:
b = B2[1, :].sum() #sum of second row of B
c = B2[:, 1] * A2[:, 0]#second column of B times first column of A:
d = c.max()#max of c
e_vector = A2[0, :] / B2[:3, 2]#e_vector: first column of A divide first three rows from the third column B
e = e_vector.sum()#e = sum
```
