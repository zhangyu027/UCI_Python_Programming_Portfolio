```python
loan_amount = 800_000
annual_interest_rate = 5
loan_years = 30
months = loan_years * 12
monthly_rate = annual_interest_rate / 1200

monthly_payment = monthly_rate * loan_amount / (1 - (1 + monthly_rate) ** (-months))
total_paid = monthly_payment * months
total_interest = total_paid - loan_amount

balance = loan_amount
for month in range(1, months + 1):
    interest_paid = balance * monthly_rate
    principal_paid = monthly_payment - interest_paid
    balance -= principal_paid

print(f'Monthly payment = ${monthly_payment:,.2f}')
print(f'Total interest amount paid by a customer in 30 years = ${total_interest:,.2f}')
print(f'Final loan balance after 30 years = ${balance:,.2f}')
print('Loan balance is approximately zero:', abs(balance) < 0.01)
```

```python
# Compute monthly payment

t = (1+interestRateMonthly)**(-durationMonths)
print('The value of t = ',t)

monthlyPayment = (loanAmount * interestRateMonthly)/(1-t)
print('Monthly Payment = ',monthlyPayment)

balance = loanAmount
```

#### Write a program which will find all such numbers which are divisible by 7 but are not a multiple of 5, between 2000 and 3200 (both included). The numbers obtained should be printed in a comma-separated sequence on a single line.

```python
import numpy as np
```

```python
l=[]
for n in range(2000,3200):
    if (n%7 == 0 and n%5 != 0):
        l.append(n)

print(*l, sep=',')
```

## Problem 3

#### With a given integral number n, write a program to generate a dictionary that contains (i, i*i) such that is an integral number between 1 and n (both included). and then the program should print the dictionary.

```python
# Method#1

n = int(input())
out_dict = {j:j*j for j in np.arange(1,n+1)}
print(out_dict)
```
