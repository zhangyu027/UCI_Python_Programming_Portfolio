## Problem 2. Write a for loop that is very similar to problem 1; it will print the numbers 0 through 9. It will also print ‘zero’, ‘even’ or ‘odd’ on each loop, depending on if the number it is printing is ‘zero’, ‘even’, or ‘odd’. For example, on the first loop, it should print 0 and ‘zero’; on the second loop, it should print 1 and ‘odd’; on the third loop, it should print 2 and ‘even’. Use the modulo operator to test if a number is even. The modulo operator in python is ‘%’. The modulo divides one number by the other and returns the remainder.  Here are some examples of using the modulo operator:

```python
for number in range(10):
    if number == 0:
        label = 'zero'
    elif number % 2 == 0:
        label = 'even'
    else:
        label = 'odd'
    print(number, label)
```

```python
for num in range(10):
    if num == 0:
        print(num, 'zero')
    elif num % 2 == 0:
        print(num, 'even')
    else:
        print(num, 'odd')
```

## Problem 3. Create a variable named my_var with the value 2. Then, write a while loop that multiplies the value of my_var by a factor of 1.65 on each loop. The while loop will continue while my_var is less than 100. Then, after the while loop completes, print the value of my_var.

```python
my_var = 2
while my_var < 100:
    my_var *= 1.65
print(my_var)
```

## Problem 5.Write a for loop that loops through a list of dictionaries. The code inside the loop will check if the dictionary contains the key ‘name’. If the dictionary does contain the key ‘name’, then the code will print the value of that key. For example, if the list of dictionaries is the list defined below, the loop would print ‘Fraiser’ and ‘Cindy’.

```python
person_1 = {'name': 'Fraiser', 'age': 52}
person_2 = {'name': 'Cindy', 'age': 46}
person_3 = {'name': 'Tom','age': 28}
people_list = [person_1, person_2, person_3]

for person in people_list:
    if 'name' in person:
        print(person['name'])
```
