```python
df = pd.read_csv('M6_Pandas_Groupby_example_Data.csv')
df.head(6)
```

```python
df.groupby('Dept').count()['Budget']
```

```python
df.groupby('Dept').sum()['Budget']
```

```python
# df.groupby('Dept').mean()['Budget']
```

```python
from numpy import int64
# Convert Budget to numeric; invalid values become NaN.
df['Budget'] = pd.to_numeric(df['Budget'], errors='coerce').astype(float)

# Group rows by department and calculate the average Budget for each department.
df.groupby('Dept')['Budget'].mean()
```
