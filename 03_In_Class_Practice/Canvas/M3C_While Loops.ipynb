# While Loops
While loops are like for loops, except they repeat a code block while a certain condition is true. Let's see some examples below.

```python
# Here is an exmple
myNum = 2
while myNum < 100:
    print(myNum)
    myNum = myNum * 1.65
```

```python
# In this example, we use a while loop to remove leading zeros from a list
initialList = [0, 0, 0, 0, 5, 6, 1, 3]
while initialList[0] == 0:
    del initialList[0]
    print(initialList)
print(initialList)
```

1. Infinite Loop with break

```python
while True:
    name = input("Enter your name (or 'quit' to exit): ")

    if name.lower() == "quit":
        break

    print("Hello,", name)

print("Program ended.")
```

2. Sentinel-Controlled Loop

```python
numbers = []
value = int(input("Enter a number (-1 to stop): "))

while value != -1:
    numbers.append(value)
    value = int(input("Enter a number (-1 to stop): "))

print("Numbers entered:", numbers)
```

3. Input Validation

```python
age = -1

while age < 0:
    age = int(input("Enter your age: "))

print("Your age is", age)
```

```python
# In this cell, please write a couple while loops.

# -------------------------------------
# Example 1: Count from 1 to 10
# -------------------------------------
count = 1

while count <= 10:
    print(count)
    count += 1

print("Finished counting!\n")
```

```python

# -------------------------------------
# Example 2: Countdown from 10 to 1
# -------------------------------------
count = 10

while count >= 1:
    print(count)
    count -= 1

print("Blast Off!\n")
```

```python
# -------------------------------------
# Example 3: Print even numbers from 2 to 20
# -------------------------------------
num = 2

while num <= 20:
    print(num)
    num += 2

print()
```

```python

# -------------------------------------
# Example 4: Find the sum of numbers 1 to 100
# -------------------------------------
number = 1
total = 0

while number <= 100:
    total += number
    number += 1

print("Sum =", total)
print()
```

```python

# -------------------------------------
# Example 5: Print powers of 2
# -------------------------------------
value = 1

while value <= 128:
    print(value)
    value *= 2

print()
```

```python
# -------------------------------------
# Example 6: Iterate through a list
# -------------------------------------
fruits = ["Apple", "Banana", "Orange", "Mango"]

index = 0

while index < len(fruits):
    print(fruits[index])
    index += 1

print()
```

```python

# -------------------------------------
# Example 7: Find the first number divisible by 17
# -------------------------------------
number = 1

while number <= 100:
    if number % 17 == 0:
        print("First multiple of 17:", number)
        break
    number += 1

print()
```

```python
# -------------------------------------
# Example 8: Skip multiples of 3
# -------------------------------------
number = 1

while number <= 15:
    if number % 3 == 0:
        number += 1
        continue

    print(number)
    number += 1

print()
```

```python

# -------------------------------------
# Example 9: Password Example
# -------------------------------------
password = ""

while password != "python123":
    password = input("Enter password: ")

print("Access Granted!")

```

```python

# -------------------------------------
# Example 10: Guess the Secret Number
# -------------------------------------
secret = 8
guess = 0

while guess != secret:
    guess = int(input("Guess the secret number (1-10): "))

print("Congratulations! You guessed correctly.")
```
