# Importing Your Own Modules

## Learning Objective
Learn how to create a Python module, import it, and call a function from it.

A **module** is a Python file containing reusable code.  
In this example, the module will be named `my_functions.py`.


## Step 1: Create the module file

The next cell creates that file in the notebook's current working folder.


````python
%%writefile my_functions.py
def my_sum(numbers):
    """Return the sum of all values in a list of numbers."""
    return sum(numbers)
````

## Step 2: Import the module

After the file has been created, Python can import it by using the file name without `.py`.


````python
import my_functions
````

## Step 3: Use a function from the module

Use dot notation:

`module_name.function_name(arguments)`


````python
answer = my_functions.my_sum([1, 2, 3])
print(answer)
````

## Expected Result

The output should be:

```text
6
```

The list contains `1`, `2`, and `3`, so the function returns their sum.


## Optional: Test the module with another list

Changing the input confirms that the same function can be reused.


````python
second_answer = my_functions.my_sum([10, 20, 30])
print(second_answer)
````

## Key Point

A custom module must be available in the current folder or on Python's import path before it can be imported.
