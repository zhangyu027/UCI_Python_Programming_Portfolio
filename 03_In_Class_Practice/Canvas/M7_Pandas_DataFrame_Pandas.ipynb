## Loading data into Pandas

```python
import pandas as pd
```

## Data is given column wise in a dictionary

```python
grades_dict = {'Wally': [87, 96, 70], 'Eva': [100, 87, 90], 'Sam': [94, 77, 90], 'Katie' : [100, 81, 82], 'Bob' : [83, 65, 85]}
```

##1. Creating a DataFrame from a Dictionary

```python
grades = pd.DataFrame(grades_dict)
print(grades)
```

```python
grades.index = index = ['Test1', 'Test2', 'Test3']
grades
```

```python
grades.T
```

##2. Customizing a DataFrame's indices with the index Attribute

```python
grades = pd.DataFrame(grades_dict, index = ['Test1', 'Test2', 'Test3'])
print(grades)
```

##3. Accessing a DataFrame's Columns

```python
grades['Eva']
```

```python
grades.Sam
```

##4. Selecting Rows  via loc and iloc Attributes

```python
grades.loc['Test1']
```

```python
grades.iloc[1]
```

##5. Selecting Rows via Slices and Lists with the loc and iloc Attributes

```python
grades.loc['Test1':'Test3']
```

```python
grades.iloc[0:2]
```

```python
grades.loc[['Test1','Test3']]
```

```python
grades.iloc[0,2]
```

```python
grades.iloc[[0,2]]
```

##6. Selecting Subsets of the Rows and Columns

```python
grades.loc['Test1':'Test2',['Eva', 'Katie']]
```

```python
grades.iloc[[0,2],0:3]
```

##7. Boolean Indexing

```python
grades[grades >= 90]
```

```python
grades[(grades >= 80) & (grades < 90)]
```

##8. Descriptive Statistics

```python
grades.describe()
```

```python
grades.describe()
```

```python
grades.mean()
```

##9. Transposing the DataFrame with T Attribute

```python
grades.T
```

```python
grades.T.describe()
```

```python
grades.T.mean()
```

##10. Sorting by Rows by Their Indices

```python
grades.sort_index(ascending=False)
```

##11. Sorting by Column Indices

```python
grades.sort_index(axis=1)
```

##12. Sorting by Column Values

```python
grades.sort_values(by='Test1', axis=1, ascending=False)
```

```python
grades.T.sort_values(by='Test1', ascending=False)
```

```python
grades.loc['Test1'].sort_values(ascending=False)
```
