# Introduction to Booleans and Comparison Operators

Booleans are simple True and False values. They are usually the result of a comparison operator, as we will see below.

## Boolean examples

```python
my_bool = False
my_other_bool = True
print(my_bool)
print(my_other_bool)
print(type(my_bool))
```

# Introduction To Comparison Operators
Comparison operators are what we use to generate booleans.  They also help determine how our script or program flows - meaning if a certain value is positive, our script may do one thing, if it is negative, it may do something else.

For example, you may want to read in transaction data, and write out the transactions that are above a certain amount, you need comparison operators to determine if the transaction amount is greater than or less than your threshold value.

Let's see some examples below:

```python
# We can use == to check if two values are equal:
print('Sophie' == 'Sophie')

print(not('Sophie' == 'Sophie'))

# We can check if two values are not equal, using !=
print('Sophie' != 'Beth')

# We can check if a value is greater than, or less than, using > or <
print(5 > 7)
print(9 < 10)

# We can also use 'greater than or equal to, >=, or 'less than or equal to' <=
print(2 <= 3)
```

## And / Or
We can also check if two thing are true, and we check if one of two (or more things) are true. We do this by using the 'and' and 'or' keywords.

```python
my_boolean = (2 > 1) and (7 > 6)
print(my_boolean)

my_boolean = (-5 < 1) or (7 < 2)
print(my_boolean)
```

## Using the in operator with lists, dictionaries, and sets

```python
# Check if an item is in a list
list_1 = ['a' , 'b', 5]
item_in_list = 'b' in list_1
print(item_in_list)


# Check if a key is in a dictionary
user_info = {'name': 'Jill Smith', 'age': 35, 'state': 'OR'}
key_in_dict = 'name' in user_info
print(key_in_dict)


# Check if item is in set
my_set = {'apple', 'banana', 'orange'}
item_in_set = 'lemon' in my_set
print(item_in_set)
```

```python
user_info = {'name': 'Jill Smith', 'age': 35, 'state': 'CA'}

customer_info = {'name': 'Sophie', 'age': 25, 'state': 'sc'}

user_indic = 'name' in customer_info
print(user_indic)
```

## A empty list converts to a False value, and a non-empty list to a True

```python
list1 = []
print(bool(list1))

list2 = [2, 3]
print(bool(list2))
```
