```python
import pandas as pd
df = pd.read_csv('M8_world-happiness-report-2021.csv')
```

```python
#Page 191

fig, ax = plt.subplots()

ax.bar(df['Regional indicator'], df['Freedom to make life choices'], label='Freedom score')
ax.legend()
plt.show()
```

```python
#Page 191

fig, ax = plt.subplots()

ax.bar(df['Regional indicator'], df['Freedom to make life choices'], label='Freedom score')
#ax.legend()

plt.xticks(rotation=45, wrap=False)
plt.show()
```

```python
#Page 191

fig, ax = plt.subplots()

ax.bar(df['Regional indicator'], df['Freedom to make life choices'], label='Freedom score')
#ax.legend()

plt.xticks(rotation=90, wrap=False)
plt.show()
```

```python
print(df['Freedom to make life choices'].describe())
```

```python
# Page 192

mean = df['Freedom to make life choices'].mean()
median = df['Freedom to make life choices'].median()

fig, ax = plt.subplots()
ax.bar(df['Regional indicator'], df['Freedom to make life choices'], label='Freedom score')
plt.xticks(rotation=90, wrap=False)

ax.axhline(mean,0,1, color='red', label='Mean Value for Dataset')
ax.axhline(median,0,1, color='green', label='Median Value for Dataset')

#ax.legend()
plt.show()
```

```python
df_means = df.groupby('Regional indicator')['Freedom to make life choices'].mean()
#print(df_means)
```

```python
fig, ax = plt.subplots()

# colors = np.arange(len(df['Regional indicator']))
# print(colors)

colors = ['red','green','blue','black','yellow','blue']

ax.bar(df_means.index, df_means, label = 'Freedom score', color = colors)

plt.xticks(rotation=90, wrap=False)
plt.show()
```

```python
#page 199

fig, ax = plt.subplots()

colors = ['red','green','blue','black','yellow','blue']

ax.barh(df_means.index,  df_means, label = 'Freedom score', color = colors)

plt.xticks(rotation=90, wrap=False)

ax.axvline(mean,0,1, color='red', label='Mean Value for Dataset')
ax.axvline(median,0,1, color='green', label='Median Value for Dataset')

#ax.legend()
plt.show()
```
