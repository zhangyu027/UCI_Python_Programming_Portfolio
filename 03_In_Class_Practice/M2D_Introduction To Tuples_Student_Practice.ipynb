# Introduction To Tuples
Tuple are like lists except they are immutable meaning you can not change them once you have created them. Tuples are created in almost the exactly same was as lists; we just use () instead of [].

```python
# First let's make a tuple!
my_tuple = (5, 'Apples', 'California')
print(my_tuple)
```

```python
my_list = list(my_tuple)
print(my_list)
print(f"Type of my_list: {type(my_list)}")
```
