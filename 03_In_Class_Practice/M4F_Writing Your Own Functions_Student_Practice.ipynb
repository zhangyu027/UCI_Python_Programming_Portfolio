## Positional Arguments and Keyword Arguments

```python
def myFunction(arg1, arg2, mathType='div'):
    if mathType == 'div':
        answer = arg1/arg2
    elif mathType == 'plus':
        answer = arg1 + arg2
    return answer
```

```python
myFunction(arg1=1, arg2=2, mathType='div')
```

```python
myFunction(arg1=1, arg2=2, mathType='plus')
```

## Functions, will look for variables first in their scope, and then outside their scope

```python
base = 10
```

```python
def plusSomething(arg1):
    answer = base + arg1
    return answer
```

```python
plusSomething(5)
```
