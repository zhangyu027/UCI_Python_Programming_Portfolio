## build up sum list funciton


```python
def sum_list(my_list):
    total = 0
    for num in my_list:
        total = total + num
    return total
```

```python
num_list = [1, 5, 10, 2]
result = sum_list(num_list)
print(result)
```

## count the number of 1s from the list

```python
def count1s(number_list):
    count = 0
    for num in number_list:
        if num == 1:
            count = count + 1
    return count
```

```python
num_list  = [1, 2, 1, 2, 1, 1, 1]
count1s(num_list)
```

## StringEvenOdd

```python
def StringEvenOdd(String):
    if len(String) %2 == 0:
        evenodd = "Even"
    else:
        evenodd = "Odd"
    return evenodd
```

```python
print(StringEvenOdd("California"))
```

```python
print(StringEvenOdd('I am a good guy'))
```

```python
df = pd.read_csv("M8_world-happiness-report-2021.csv")
```

```python
fig, ax = plt.subplots(figsize=(10,6))

ax.bar(df['Regional indicator'], df['Freedom to make life choices'], label='Freedom score')
ax.legend()
plt.xticks(rotation=45)
plt.show()
```

## 2 pie charts

```python
fig, ax = plt.subplots(figsize=(10,6))

# Group by 'Regional indicator' and sum 'Freedom to make life choices'
regional_freedom = df.groupby('Regional indicator')['Freedom to make life choices'].sum()

# Plot the pie chart
ax.pie(regional_freedom.values, labels=regional_freedom.index, autopct='%1.1f%%', startangle=90)
ax.set_title('Distribution of Freedom to Make Life Choices by Regional Indicator')
ax.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.

plt.show()
```
