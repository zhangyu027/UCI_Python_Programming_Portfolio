## The top of the die is given by $3(a^3b - ab^3) \bmod 7$. Compute for (a) a=2, b=6 and (b) a=3, b=5.

```python
def die_top(a, b):
    return (3 * (a**3 * b - a * b**3)) % 7

case_a = die_top(2, 6)
case_b = die_top(3, 5)
case_a, case_b
```

##A sheet of paper has thickness $t=0.1$ mm. How many folds are needed to reach the Moon, distance $d=384,400$ km? After $n$ folds, thickness is $2^n t$.

```python
import math

t_mm = 0.1
d_km = 384_400
d_mm = d_km * 1_000_000  # 1 km = 1,000,000 mm

n = math.ceil(math.log2(d_mm / t_mm))
thickness_after_n_mm = (2 ** n) * t_mm
thickness_after_n_km = thickness_after_n_mm / 1_000_000

n, thickness_after_n_km
```

## Explain how the incorrect tests are interpreted by Python and give a correct alternative.

```python
s = 'eggs'

# Incorrect examples:
example_1 = s == ('eggs' or 'ham')   # ('eggs' or 'ham') evaluates to 'eggs'
example_2 = s == ('ham' or 'eggs')   # ('ham' or 'eggs') evaluates to 'ham'

# Correct alternatives:
correct_1 = (s == 'eggs') or (s == 'ham')
correct_2 = s in ('eggs', 'ham')

example_1, example_2, correct_1, correct_2
```
