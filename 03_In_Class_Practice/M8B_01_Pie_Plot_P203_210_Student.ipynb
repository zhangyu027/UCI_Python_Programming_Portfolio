```python
#1
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [15, 25, 25, 30, 5]

fig, ax = plt.subplots()
ax.pie(x)
plt.show()
```

```python
x = [15, 25, 25, 30, 5]

labels = ['Very Likely', 'Likely', 'Unsure', 'Unlikely', 'Very Unlikely']

fig, ax = plt.subplots()
ax.pie(x, labels = labels)
ax.set_title('Survey responses')
plt.show()
```

```python
x = [15, 25, 25, 30, 5]

labels = ['Very Likely', 'Likely', 'Unsure', 'Unlikely', 'Very Unlikely']
colors = ['red','green','blue','black','yellow','blue']

fig, ax = plt.subplots()
ax.pie(x, labels = labels, colors = colors)
ax.set_title('Survey responses')
plt.show()
```

```python
x = [15, 25, 25, 30, 5]

labels = ['Very Likely', 'Likely', 'Unsure', 'Unlikely', 'Very Unlikely']
colors = ['red','green','blue','black','yellow','blue']

fig, ax = plt.subplots()
ax.pie(x, labels = labels, colors = colors, autopct='%.0f%%')
ax.set_title('Survey responses')
plt.show()
```

```python
x = [15, 25, 25, 30, 5]

labels = ['Very Likely', 'Likely', 'Unsure', 'Unlikely', 'Very Unlikely']
colors = ['red','green','blue','black','yellow','blue']
explode = [0,0,0,0,0.2]

fig, ax = plt.subplots()
ax.pie(x, labels = labels, colors = colors, autopct='%.0f%%', explode = explode)
ax.set_title('Survey responses')

plt.show()
```
