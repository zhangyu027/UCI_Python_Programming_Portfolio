Problem 5: mortgage application.

```python
loan_amount = 800_000
annual_interest_rate = 5
loan_years = 30
months = loan_years * 12
monthly_rate = annual_interest_rate / 1200

monthly_payment = monthly_rate * loan_amount / (1 - (1 + monthly_rate) ** (-months))
total_paid = monthly_payment * months
total_interest = total_paid - loan_amount

balance = loan_amount
for month in range(1, months + 1):
    interest_paid = balance * monthly_rate
    principal_paid = monthly_payment - interest_paid
    balance -= principal_paid

print(f'Monthly payment = ${monthly_payment:,.2f}')
print(f'Total interest amount paid by a customer in 30 years = ${total_interest:,.2f}')
print(f'Final loan balance after 30 years = ${balance:,.2f}')
print('Loan balance is approximately zero:', abs(balance) < 0.01)
```
