## Problem 2. Select ‘Title’ and ‘Year’ of the ‘table1’, into another table and name is ‘table2’.

```python
table2 = table1[["Title", "Year"]]
table2
```

## Problem 3. Select only the movies from ‘table1’ where the gross revenues generated by the movie exceeds 100 million.

```python
gross_over_100m = table1[table1["Gross"] > 100000000]
gross_over_100m
```

```python
gross_over_100m.sort_values("Gross", ascending=True)
```

## Problem 4.  Select only the movies from ‘table1’ where the gross revenues generated by the movie is between 100 and 200 million. Sort the output data frame by 'Gross' revenues in descending order. Display only the top 10 entries.

```python
gross_100_to_200m_top10 = (
    table1[(table1["Gross"] >= 100_000_000) & (table1["Gross"] <= 200_000_000)]
    .sort_values("Gross", ascending=False)
    .head(10)
)
gross_100_to_200m_top10
```

## Problem 5. Select movies where the word ‘the’ appears in the title of the movie.  Show the first 10 movies.

```python
table1[table1['Title'].str.contains('the', na=False)].head(10)
```

```python
movies_with_the = table1[table1["Title"].str.contains("the", case=False, na=False)]
movies_with_the.head(10)
```

## Problem 6. Sort the movies in ‘table1’ in descending order (highest value first) by their ‘Gross’ revenues.  Select top 3 movies made the most amount of money based on ‘Gross’ column? Select top 3 movies made the most amount of money based on ‘Gross (Adjusted)’ column?

```python
top3_gross = table1.sort_values("Gross", ascending=False).head(3)
top3_adjusted = table1.sort_values("Gross (Adjusted)", ascending=False).head(3)

print("Top 3 by Gross")
display(top3_gross)

print("Top 3 by Gross (Adjusted)")
display(top3_adjusted)
```

## Problem 7. Count the number of movies from each studio.  How many movies made by ‘Columbia’ and ‘Fox’ studios?

```python
studio_counts = table1["Studio"].value_counts()
display(studio_counts)

print("Columbia:", studio_counts.get("Columbia", 0))
print("Fox:", studio_counts.get("Fox", 0))
```

## Problem 8. Which is the highest grossing movie from ‘Columbia’ studio?

```python
highest_columbia = table1[table1["Studio"] == "Columbia"].sort_values("Gross", ascending=False).head(1)
highest_columbia
```

```python
table1[table1["Studio"] == "Columbia"].nlargest(1, "Gross")
```

```python
table1[table1["Studio"] == "Columbia"] \
    .sort_values("Gross", ascending=False) \
    .head(1)
```

## Problem 9.  Compute the ratio between ‘Gross (Adjusted)’ and ‘Gross’ for all movies.  Next sort the table by the ratio in descending order.  Display the top 10 ratios.

```python
table1["Ratio"] = table1["Gross (Adjusted)"] / table1["Gross"]
top10_ratios = table1.sort_values("Ratio", ascending=False).head(10)
top10_ratios
```

## Problem 10. Group all the movies by the year in which they were released.  In which year has the maximum number of movies released?

```python
movies_by_year = table1.groupby("Year").size().sort_values(ascending=False)
display(movies_by_year)

print("Year with maximum number of movies:", movies_by_year.idxmax())
print("Number of movies:", movies_by_year.max())
```
