```python
import pandas as pd

temps = {'Mon': [68, 89], 'Tue': [71, 93], 'Wed': [66, 82], 'Thu': [75, 97], 'Fri': [62, 79]}
```

Problem 1.Convert the Python dictionary ‘temps’ into the Python DataFrame named ‘temperatures’ with ‘Low’ and ‘High’ as the indices, then display the DataFrame.

```python
temperatures = pd.DataFrame(temps, index=['Low', 'High'])
temperatures
```

Problem 2. Use the column names to select only the columns for ‘Mon’ through ‘Wed’.

```python
temperatures.loc[:, 'Mon':'Wed']
```

Problem 3. Use the row index ‘Low’ to select only the low temperatures for each day.

```python
temperatures.loc['Low']
```

Problem 4. Set the floating-point precision to 2, then calculate the average temperature for each day.

```python
pd.set_option('display.precision', 2)
temperatures.mean(axis=0)
```

Problem 5. Calculate the average low and high temperatures.

```python
temperatures.mean(axis=1)
```
