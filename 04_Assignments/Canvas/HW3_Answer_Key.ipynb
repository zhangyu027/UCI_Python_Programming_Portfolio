## Problem 2
Print the numbers **0 through 9** and classify each number as `zero`, `even`, or `odd`.

Use the modulo operator `%` to determine whether a nonzero number is even or odd.


````python
for number in range(10):
    if number == 0:
        print(number, "zero")
    elif number % 2 == 0:
        print(number, "even")
    else:
        print(number, "odd")
````

**Expected output**
```text
0 zero
1 odd
2 even
3 odd
4 even
5 odd
6 even
7 odd
8 even
9 odd
```


## Problem 3
Create `my_var` with an initial value of `2`. While it is less than `100`, multiply it by `1.65`.  
After the loop finishes, print the final value.


````python
my_var = 2

while my_var < 100:
    my_var = my_var * 1.65

print(my_var)
````

The final value will be the first multiplied value that reaches or exceeds `100`.

Because floating-point values are used, Colab/Python will display a decimal value.


## Problem 4
Determine `shipping_cost` from `package_weight`:

- Weight ≤ 5 → cost = 3
- Weight > 5 and ≤ 10 → cost = 7
- Weight > 10 → cost = 15

The assignment asks students to test the code with different package weights.


````python
package_weight = 1

if package_weight <= 5:
    shipping_cost = 3
elif package_weight <= 10:
    shipping_cost = 7
else:
    shipping_cost = 15

print("Package weight:", package_weight)
print("Shipping cost:", shipping_cost)
````

### Problem 4 — Additional Tests
This cell makes it easy for the instructor to verify all three branches.


````python
test_weights = [1, 5, 6, 10, 11, 20]

for package_weight in test_weights:
    if package_weight <= 5:
        shipping_cost = 3
    elif package_weight <= 10:
        shipping_cost = 7
    else:
        shipping_cost = 15

    print("Weight:", package_weight, "-> Shipping cost:", shipping_cost)
````

**Expected output**
```text
Weight: 1 -> Shipping cost: 3
Weight: 5 -> Shipping cost: 3
Weight: 6 -> Shipping cost: 7
Weight: 10 -> Shipping cost: 7
Weight: 11 -> Shipping cost: 15
Weight: 20 -> Shipping cost: 15
```


## Problem 5
Loop through a list of dictionaries. If a dictionary contains the key `name`, print the value associated with that key.


````python
person_1 = {"name": "Fraiser", "age": 52}
person_2 = {"name": "Cindy", "age": 46}

people_list = [person_1, person_2]

for person in people_list:
    if "name" in person:
        print(person["name"])
````

**Expected output**
```text
Fraiser
Cindy
```


### Problem 5 — Optional Instructor Test
The assignment requires printing a name only when the `name` key exists.  
This extra test demonstrates that a dictionary without `name` is safely skipped.


````python
person_1 = {"name": "Fraiser", "age": 52}
person_2 = {"name": "Cindy", "age": 46}
person_3 = {"age": 30}

people_list = [person_1, person_2, person_3]

for person in people_list:
    if "name" in person:
        print(person["name"])
````
